```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

```

```python
df= pd.read_csv("../DATA/Salary Data.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
input=df[["YearsExperience"]]  ## sklearn need multidimenssion arrary
input.ndim
output = df['Salary']
sns.scatterplot(x="YearsExperience", y="Salary", data=df)

```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(input,output, train_size=.25)
```

```python
from sklearn.linear_model import LinearRegression
lr= LinearRegression()

lr.fit(x_train,y_train)
```

```python
lr.score(x_test,y_test)*100
```

```python
lr.predict([[7]])
```

```python
## y=m*x+c
lr.coef_  # coefficient thia 0 less tha 90
```

```python
lr.intercept_ # intercenpt 
y_prdict= lr.predict(df[["YearsExperience"]])
```

```python
sns.scatterplot(x=df["YearsExperience"],y=df["Salary"])
plt.plot(df["YearsExperience"],y_prdict,c="red")
plt.legend(["org data ","predict line"])
plt.show()
```
